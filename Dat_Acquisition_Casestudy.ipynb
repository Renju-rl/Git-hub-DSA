#Q1

```python
import requests
import sqlite3
import numpy as np
import pandas as pd
url = ' https://api.spacexdata.com/v4/launches'
```

##Extract relevant columns: name, date_utc, success, details, rocket

```python

response = requests.get(url)
data = response.json()

df_data = pd.DataFrame(data)

df_data = df_data[['name', 'date_utc', 'success', 'details', 'rocket']]

df_data.head()
```

##Convert date_utc to datetime and extract the year

```python
df_data['date_utc'] = pd.to_datetime(df_data['date_utc'])
df_data['year'] = df_data['date_utc'].dt.year

df_data.head()
```

#Q2

##Extract id, name, type, active, and stages


```python
url = 'https://api.spacexdata.com/v4/rockets'
response = requests.get(url)
data2 = response.json()
df_data2 = pd.DataFrame(data2)

df_data2 = df_data2[['id', 'name', 'type', 'active','stages']]
df_data2.head()
```

#Q3

##Merge Launch and Rocket Data

*   Join the two DataFrames on rocket ID using



```python
merge_df = pd.DataFrame(df_data.merge(df_data2, left_on="rocket",right_on="id"))
merge_df.head()
```

#Q4


##Add Simulated Country Information

*   Add a new column country and randomly assign one of these values:
*   ['USA', 'Russia', 'India', 'China', 'France']



```python
countries = ['USA', 'Russia', 'India', 'China', 'France']
merge_df["country"] = np.random.choice(countries, size=len(merge_df))
merge_df
```

#Q5


*  Use sqlite3 to create a connection and save the merged DataFrame as a table named
launches


*   Table should contain all merged columns including country





```python
conn = sqlite3.connect('spacex.db')
merge_df.to_sql('launches', conn, if_exists='replace', index=False)
conn.close()
merge_df
```

#Q6

*   Launches by Country


```python
conn = sqlite3.connect('spacex.db')

query = """
SELECT country, COUNT(*) AS launch_count
FROM launches
GROUP BY country
ORDER BY launch_count DESC;
"""
df_country = pd.read_sql_query(query, conn)
df_country
```

##Which year had the highest number of launches?

```python
query = """
SELECT year, COUNT(*) AS launch_count
FROM launches
GROUP BY year
ORDER BY launch_count DESC
LIMIT 1;
"""
df_top_year = pd.read_sql_query(query, conn)
df_top_year
```

##  Top 5 Missions by Launch Count

```python
query = """
SELECT name_x, COUNT(*) AS launch_count
FROM launches
GROUP BY name_x
ORDER BY launch_count DESC;
"""

df_top_missions = pd.read_sql_query(query, conn)
df_top_missions.head()
```
