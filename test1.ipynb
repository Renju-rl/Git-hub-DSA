```python
# Q1. Write a function that return the range (maximum and minimun) of a given list of numbers

def return_range(List_number):
  minimum = List_number[0] # setting the first value of the list as minimum value
  maximum = List_number[0] # setting the first value of the list as maximumu value
  for numbers in List_number:
  #looping through the entire list to check for maximum amd minimum
    if(numbers >= maximum):
      maximum = numbers
    if(numbers <= minimum):# if using here for getting this answer also
      minimum = numbers

  return(minimum,maximum)


input_list = [4, 8, 2, 9, 1, 6]
range = return_range(input_list)
range
```

```python
# Q2. Write function that returns mean value of a list of numbers
def mean_v(n):
  result = 0
  for i in n:
    result += i
  return result / len(n)

my_list=[1,2,3,4,5,6,7,8]
print(f"The mean value is: {mean_v(my_list)}")
```

```python
# Q3. Write a function that returns the median value of a list of numbers

def find_median(input_list):
  input_list = sorted(input_list)
  length_input = len(input_list)
  print(input_list)
  middle_num = length_input // 2 #Helps to find middle value of the list

  if length_input % 2 == 0:
    return (input_list[middle_num - 1] + input_list[middle_num]) / 2
  else:
    return (input_list[middle_num])


list_input = [1, 2, 3, 42, 5, 68, 66,8]
print(f"The median value is: {find_median(list_input)}")
```
